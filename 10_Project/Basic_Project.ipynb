# Calculator


```python
# validate operation input
operation = ''
while operation not in ('+', '-', '*', '/'):
    operation = input("Enter operation (+, -, *, /): ").strip()
    if operation == '':
        print("Operation cannot be empty. Please enter one of +, -, *, or /.")
    elif operation not in ('+', '-', '*', '/'):
        print("Invalid operation. Please enter one of +, -, *, or /.")

# helper to read a float safely
def read_float(prompt):
    while True:
        s = input(prompt).strip()
        if s == '':
            print("Input cannot be empty. Please enter a number.")
            continue
        try:
            return float(s)
        except ValueError:
            print("Invalid number. Please enter a valid numeric value.")

# use the safe function
num1 = read_float("Enter first number: ")
num2 = read_float("Enter second number: ")

if operation == '+':
    result = num1 + num2
    print(f"{num1} + {num2} = {result}")
elif operation == '-':
    result = num1 - num2
    print(f"{num1} - {num2} = {result}")
elif operation == '*':
    result = num1 * num2
    print(f"{num1} * {num2} = {result}")
elif operation == '/':
    if num2 != 0:
        result = num1 / num2
        print(f"{num1} / {num2} = {result}")
    else:
        print("Error: Division by zero is not allowed.")
```

# Temperature Conversion Program 

```python
unit = input("Enter 'C' to convert to Celsius or 'F' to convert to Fahrenheit: ").upper()
temp = float(input("Enter the temperature: "))

if unit == 'C':
    converted_temp = (temp - 32) * 5.0/9.0
    print(f"The temperature in Celsius is: {converted_temp} °C")
elif unit == 'F':
    converted_temp = (temp * 9.0/5.0) + 32
    print(f"The temperature in Fahrenheit is: {converted_temp} °F")
else:
    print("Invalid input. Please enter 'C' or 'F'.")
```

# Countdown Timer Program ⌛

```python
import time

my_time = int(input("Enter time in seconds for countdown: "))

for x in range(my_time, 0, -1):
    seconds = x % 60
    minutes = int(x / 60) % 60
    hours = int(x / 3600) % 24
    print(f"{hours:02}:{minutes:02}:{seconds:02}")
    time.sleep(1)
    
print("Time's up!")
```

# Shopping Cart Program 📋

```python
foods = []
prices = []
total = 0

while True:
    food = input("Enter a food to buy (q to quit): ,")
    if food == 'q':
        break
    else:
        price = float(input(f"Enter the price of {food}: $"))
        foods.append(food)
        prices.append(price)
        total += price
        
print("\n-----Your Shopping Cart: -----")

for food in foods:
    print(food, end=" ")
    
for price in prices:
    total += price
    
print(f"\nTotal Price: ${total}")
```
