### Framework
A tabular data structure with rows and columns (2-D). Similar to Excel spreadsheet.

```python
import pandas as pd

data = {"Name": ["Spongebob", "Patrick", "Squidward"],
        "Age": [30,32,48]
        }
df = pd.DataFrame(data , index=["Employee 1","Employee 2","Employee 3"])
print(df)
print(df.loc["Employee 1"])
print(df.iloc[2])

# Add a new column
df["Job"] = ["Fry Cook", "Spy", "Cashier"]
print(df)

#Add a new row
new_row = pd.DataFrame([{"Name":"Mendy", "Age":22, "Job": "Footballer"},
                        {"Name":"Roop", "Age":19, "Job": "Model"}],
                       index=["Employee 4","Employee 5"])
df = pd.concat([df, new_row])  #Concatenate used to combine multiple DataFrame
print(df)

new_row = pd.DataFrame([{"Name":"Mendy", "Age":22, "Job": "Footballer"}])
df = pd.concat([df, new_row], ignore_index=True)
print(df)
```

### Selection


```python
import pandas as pd

# Set Name as index for easier row selection
df = pd.read_csv(r"Dataset\pokemon.csv", index_col="Name")

# SELECTION BY COLUMN
print(df.index.to_list())                     # Print all Pokemon names
print(df[["Type 1"]].head(10))                # First 10 Pokemon types
print(df[["Type 1","HP"]].head(10))           # First 10 Pokemon types + HP

# SELECTION BY ROW
print(df.loc["Pikachu"])                      # Row for Pikachu
print(df.loc["Charizard":"Blastoise", ["Type 1","HP"]])  # Charizard to Blastoise
print(df.iloc[0:11])                          # First 11 rows
print(df.iloc[0:11:2, 1:3])                   # Every 2nd row from first 11, cols 1–2
```

### AGGREGATION Function
Reduces a set of values into a single summary value. Used to summarized and analyze data.
Often used with the groupby() function


```python
import pandas as pd

df = pd.read_csv(r"Dataset\top_100_test_player.csv")

# DataFrame-wide statistics
print("Mean:", df.mean(numeric_only=True))
print("Sum:", df.sum(numeric_only=True))
print("Median:", df.median(numeric_only=True))
print("Min:", df.min(numeric_only=True))
print("Max:", df.max(numeric_only=True))
print("Count:", df.count())

# Single column statistics (runs)
print("runs mean:", df["runs"].mean())
print("runs sum:", df["runs"].sum())
print("runs median:", df["runs"].median())
print("runs min:", df["runs"].min())
print("runs max:", df["runs"].max())
print("runs count:", df["runs"].count())

# Grouped statistics
group = df.groupby("country")
print("group mean:", group.mean(numeric_only=True))
print("group object:", group)
```

### Data Cleaning
The process of fixing/removing:
incomplete, incorrect or irrelevent data.
approx 75% of work done with Pandas is data cleaning

```python
import pandas as pd

df = pd.read_csv(r"Dataset\top_100_test_player.csv")

# 1. Drop irrelevant columns
df = df.drop(columns=["2_hundreds"])
print("After dropping irrelevant columns:\n", df)

# 2. Handle missing data
df = df.dropna(subset=["runs","innings"])
print("After dropping missing runs/innings:\n", df)

# 3. Fill missing values
dff = df.fillna({"runs": None, "innings": None})
print("After filling missing values:\n", dff)

# 4. Fix inconsistent values
df["country"] = df["name"].replace({"Pakistan":"Dalla"})
print("After fixing inconsistent values:\n", df.to_string())

# 5. Standardize text
df["name"] = df["name"].str.lower()
print("After standardizing text:\n", df.to_string())

# 6. Fix data types
df["hundreds"] = df["hundreds"].astype("string")   # or int/float depending on your need
print("After fixing data types:\n", df.to_string())

# 7. Remove duplicate values
df = df.drop_duplicates()
print("After removing duplicates:\n", df.to_string())
```
