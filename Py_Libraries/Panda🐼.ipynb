### Framework
A tabular data structure with rows and columns (2-D). Similar to Excel spreadsheet.

```python
import pandas as pd

data = {"Name": ["Spongebob", "Patrick", "Squidward"],
        "Age": [30,32,48]
        }
df = pd.DataFrame(data , index=["Employee 1","Employee 2","Employee 3"])
print(df)
print(df.loc["Employee 1"])
print(df.iloc[2])

# Add a new column
df["Job"] = ["Fry Cook", "Spy", "Cashier"]
print(df)

#Add a new row
new_row = pd.DataFrame([{"Name":"Mendy", "Age":22, "Job": "Footballer"},
                        {"Name":"Roop", "Age":19, "Job": "Model"}],
                       index=["Employee 4","Employee 5"])
df = pd.concat([df, new_row])  #Concatenate used to combine multiple DataFrame
print(df)

new_row = pd.DataFrame([{"Name":"Mendy", "Age":22, "Job": "Footballer"}])
df = pd.concat([df, new_row], ignore_index=True)
print(df)
```

### Filtering
Keeping the row that match a condition

```python
import pandas as pd

df = pd.read_json("top_100_test_batsmen_dataviz.json")

top_batsmen = df[df["Runs"] >= 12000]

Indian_batsmen = df[df["Country"] == "India"]

top_Indian_batsmen = df[(df["Runs"] >= 12000) & (df["Country"] == "India")]

print(top_batsmen)
print(Indian_batsmen)
print(top_Indian_batsmen)
```

### AGGREGATION Function
Reduces a set of values into a single summary value. Used to summarized and analyze data.
Often used with the groupby() function


```python
import pandas as pd

df = pd.read_json("top_100_test_batsmen_dataviz.json")

print(df.mean(numeric_only=True))
#print(df.sum(numeric_only=True))
#print(df.median(numeric_only=True))
#print(df.min(numeric_only=True))
#print(df.max(numeric_only=True))
#print(df.count())

# Single column
print(df["Runs"].mean())
print(df["Runs"].sum())
print(df["Runs"].median())
print(df["Runs"].min())
print(df["Runs"].max())
print(df["Runs"].count())

group = df.groupby("Country")

print(group.mean(numeric_only=True))
print(group)
```
