###Plot Customisation

```python
import matplotlib.pyplot as plt
import numpy as np  #use numpy cuz it is faster than python list

x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 29, 35, 25])
y2 = np.array([17, 23, 81, -15])

plt.plot(x,y1, marker= "*", markersize=10,
         markerfacecolor="red",markeredgecolor="#1cd3fc",
         linestyle="dashdot", linewidth= 2,color="orange")

plt.plot(x,y2, marker= ".", markersize=15,
         mfc="#1cd3fc", mec="blue",
         linestyle="solid", linewidth= 2,color="green")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np  #use numpy cuz it is faster than python list

x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 29, 35, 25])
y2 = np.array([17, -23, 81, -15])
y3 = np.array([75, 9, -5, -45])

line_style = dict(marker= ".",
                  markersize=15,
                  mfc="#1cd3fc",
                  mec="blue",
                  linestyle="solid",
                  linewidth= 2,
                  )

plt.plot(x,y1,color="red", **line_style)
plt.plot(x,y2,color="#1cfc45", **line_style)
plt.plot(x,y3,color="beige", **line_style)

plt.show()
```

### Label

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 29, 35, 25])
y2 = np.array([17, 23, 81, 15])
y3 = np.array([75, 9, 5, 45])

plt.title("Class size",
          fontsize=20,
          fontweight="bold",
          color="#2d4cfc")

plt.xlabel("Year",
          fontsize=15,
          fontweight="bold",
          color="red")
plt.ylabel("Student",
           fontsize=15,
           fontweight="bold",
           color="red")

plt.tick_params(axis="both",
                colors="#2dbefc")

plt.plot(x,y1)
plt.plot(x,y2)
plt.plot(x,y3)

plt.show()
```

###Grid line
grid() = Helps make plots easier to read by adding reference lines.

### Bar chart
Compare Categories of data by representing each category with a bar

```python
import matplotlib.pyplot as plt
import numpy as np

categories = np.array(["Grains", "Friuts", "Vegetables", "Protein", "Dairy", "Sweets"])
values = np.array([4,3,2,5,3,0])

plt.bar(categories, values, color="skyblue")

plt.title("Daily Consumption")
plt.xlabel("Food")
plt.ylabel("Calories")

plt.show()

```

### Pie chart
Circular chart divided into slices to show percentages of the total.

Good for visulaizing distribution among categories.

```python
import matplotlib.pyplot as plt
import numpy as np

categories = np.array(["Freshmen", "Sophomores", "Juniors", "Seniors"])
values = np.array([124,23,220,50])
colors = ["red", "yellow", "blue", "green"]
plt.pie(values,
        labels=categories,
        autopct="%1.1f%%",
        colors=colors,
        explode=[0,0,0,0.2],
        shadow=True,
        startangle=180)

plt.title("Bro Code College")

plt.show()
```

###Scatter Plot
Shows the relationship btw two variables.

Helps to identify a correlation (+, -, None)

Ex: Study hours v/s Test scores

```python
import matplotlib.pyplot as plt
import numpy as np

x1 = np.array([0,1,1,3,4,5,7,7,8]) # Hours studies
y1= np.array([50, 55, 60, 65, 69, 78, 93, 97,100]) # Grades

x2 = np.array([0,1,2,2,4,6,7,8,8]) # Hours studies
y2 = np.array([12, 15, 20, 61, 69, 68, 73, 67,70]) # Grades

plt.scatter(x1,y1, color = "red",
            alpha = 0.5,
            s = 100,
            label="Section A")

plt.scatter(x2,y2, color = "black",
            alpha = 0.2,
            s = 100,
            label="Section B")

plt.xlabel("Hours Studies")
plt.ylabel("Grades")
plt.title("Hours vs Grades")

plt.legend()

plt.show()
```

### Histogram
A visual representation of the distribution of quantitative data.

They group values into biins (intervals) and counts how many fall in each range.

```python
import matplotlib.pyplot as plt
import numpy as np

scores = np.random.normal(loc=80, scale=10, size=50)
scores = np.clip(scores, 50, 100)

plt.hist(scores, bins=10, color="lightgreen", edgecolor="Black")

plt.title("Test Scores")
plt.xlabel("Score")
plt.ylabel("Frequency of Students")

plt.show()
```

### Subplots
Figure = Entire Canvas 🖼️

Ax = A single plot (subplot) 📈

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1,2,3,4,5])

figure , axes = plt.subplots(2, 2)

axes[0,0].plot(x, x*2, color="red")
axes[0,0].set_title("x*2")

axes[0,1].bar(x, x**2, color="green")
axes[0,1].set_title("x**2")

axes[1,0].plot(x, x**3, color="blue")
axes[1,0].set_title("x**3")

axes[1,1].plot(x, x**4, color="orange")
axes[1,1].set_title("x**4")

plt.tight_layout()
plt.show()
```

### Pandas + Matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df = pd.read_csv("pokemon_100.csv")

type_count = df["Type1"].value_counts(ascending=True)

print(type_count)

plt.bar(type_count.index, type_count.values, color= "#03dffc", alpha=0.7)

plt.xlabel("Type")
plt.ylabel("Count")
plt.title("Pokemon Types")

plt.show()
```
